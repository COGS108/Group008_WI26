## Background and Prior Work

Our topic is on housing prices & drug usage in Portland, specifically before and after the 2020 Oregon Ballot Measure 110 was passed. Measure 110 reduced the penalties for drug possession for small amounts of controlled substances, reducing it from a class A misdemeanor to class E. We want to study how this may have affected the general population of Portland's access to housing, and if drug usage was related during this time period. By 2024 measure 110 was largely repealed.<a name = "Drug_Decriminalization">[</a><sup>4</sup>](#Drug_Decriminalization_ref)

A number of studies were conducted about this time period, all which studied the effect of drug decriminalization in Portland in specific categories. 

One study found that housing instability and opioid related harms often rise together, especially in areas facing rapid policy shifts. Using cross-sectional survey data from unstably housed individuals with a history of drug use, the study identified key demographic and behavioral factors associated with access to housing assistance, including frequency of opioid use and community supervision status. The study stated that, "Participants were recruited [...] with partner agencies that provide supportive services [...] direct outreach by the study team to homeless"<a name="cite_ref-1"></a>[<sup>2</sup>](#cite_note-1). While this research focuses on individuals already experiencing housing instability, our study differs by examining broader population-level trends in Portland before and after the passage of Measure 110. Rather than relying on survey data, we analyze changes in housing prices alongside drug usage indicators to explore whether decriminalization coincided with shifts in housing accessibility and market conditions. This allows us to investigate potential associations at a city-wide level and assess how policy changes may have impacted both housing dynamics and drug use simultaneously over time.

Another study examined whether Oregon’s 2021 decriminalization of drug possession was linked to changes in overdose mortality while accounting for the spread of fentanyl in the unregulated drug market. The study found that “decriminalization of drug possession was not associated with an increase in fatal drug overdose rates in Oregon in the 2 years after its enactment” once fentanyl was considered as a confounding factor<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2). The authors emphasized that fentanyl is “the principal driver of the overdose mortality epidemic in the US,” suggesting that increases in overdose deaths cannot be attributed to Measure 110 alone. While this research focuses on overdose mortality rather than housing outcomes, it is still relevant to our study because it highlights the importance of accounting for external factors when evaluating the effects of drug policy changes. Our research builds on this idea by examining housing prices and drug usage trends in Portland before and after Measure 110 to better understand how policy changes may have impacted the broader community.

## Hypothesis


We hypothesize that there will be a negative correlation between opioid-related incidents and both median housing prices and market velocity (average time spent on Zillow) in Portland between 2020 and 2021. Specifically, as opioid overdoses and hospitalizations increased in certain zip codes, we predict that home sale prices in those areas decreased relative to the city average, while the "days on Zillow" significantly increased.

**Combining the datasets:** Housing data is at ZIP level; overdose data is at county level. We use a county–ZIP crosswalk (`county_zip_crosswalk.csv`) to map each housing record's ZIP to its county, then join to county-year overdose counts/rates. Now we want to further dive into the pre and post differences of Measure 110. We note two things: Measure 110 was passed in November 2020, and went into effect Feburary 1st, 2021. In our earlier data cleaning, we saved the CSV priceHistory, which gave the priceHistory of every property that was taken from the Zillow API and sold between July 2020 and July 2021. We will import this data, and look to merge it with our housing data set, and finally make explore some data trends regarding the data by zip code.

```python
priceHistory = pd.read_csv('data/02-processed/priceHistory.csv')
priceHistory.head()
```

We first note how our data is presented. These are index/row based similarity, so 19309 NE Glisan St, for instance, was on Zillow for 25 days. Therefore, we're going to want to use .join to combine our datasets. We will also drop the description, dateSold, daysOnZillow, lastSoldPrice, yearSold, and month_year_sold columns from the housing data frame, since these are either irrevalent or duplicates information.

```python
housing_drop_cols = housing[['abbreviatedAddress','address/city','address/zipcode','bathrooms','bedrooms','livingArea','price']]
price_history_housing = housing_drop_cols.join(priceHistory, how='left', lsuffix='_left', rsuffix='_right')
price_history_housing
```

First, we investigate pre Measure 110. We will use the dateSold column as a means to obtain our pre-110 dataset, aptly named "pre_110". We will want to get summary statistics on the priceing of houses, as we as group by zipcodes to explore trends in certain subsects of Portland. To do this, we first convert the **object** datatype in dateSold to a **datetime** type. Then, we will extract the month and year of each **datetime**, check if it is in the pre Measure 110 range we want to dive into, add that row to our pre_110 dataframe, and generate summary statistics.

```python
price_history_housing['dateSold'] = pd.to_datetime(price_history_housing['dateSold'])
pre_110 = price_history_housing[(price_history_housing['dateSold'].dt.year == 2020) & (price_history_housing['dateSold'].dt.month < 12)]
print(pre_110.shape)
pre_110.head()
```

We're given 4700 datapoints to work with. We acknowledge that some prices will be far lower than the mean test statistic it generates, and some much higher, but it opens the possibility of looking at opioid incidences in these regions later in our study, so we will not filter by price. First, we generate a general summary statistic on the whole price column.

```python
pre_110['price'].describe()
```

Now, group by zipcodes. First, we figure out how many unique zip codes there are with .unique, and then use .describe() on the price column with zip codes filtered. This is going to be a lot of summary statistics printed!

```python
zipcodes_pre = pre_110['address/zipcode'].unique()
for zipcode in zipcodes_pre:
    print('-- SUMMARY FOR ' + str(zipcode) + ' --')
    print(pre_110[pre_110['address/zipcode'] == zipcode]['price'].describe())
```

On further examination, we see some zipcodes have a miniscule amount of datapoints related to them. We will only view the top 10 of these datapoints. Also, based on the range of price we had in our summary statistics, we will cap things to be only within 98% range of our data; this exludes the bottom 1%, and top 1%. Now, we will want to visualize the data for each of these zipcodes. We have chosen to do a box plot, as it captures the spread of the dataset the most, which is what we are invested in for comparing our ranges. We will also include mean as a measure of central tendancy by setting the showmeans flag to be true. It is almost sure that the mean will trend above the median price, since we have more outliers in the millions of dollars range, rather than, say, 5 digit or even 4 digit housing prices.

```python
lower = pre_110['price'].quantile(0.1)
upper = pre_110['price'].quantile(0.99)
plt.figure(figsize=(14,6))

top_zips_pre = pre_110['address/zipcode'].value_counts().head(10).index
pre_filtered = pre_110[pre_110['address/zipcode'].isin(top_zips_pre)]
pre_filtered = pre_filtered[(pre_filtered['price'] >= lower) & (pre_filtered['price'] <= upper)]

print(pre_filtered.shape)

sns.boxplot(x='address/zipcode', y='price', hue='address/zipcode', data=pre_filtered, showmeans=True, meanprops={'marker': 'o', 'markerfacecolor': 'white', 'markeredgecolor': 'black', 'markersize': '8'})

plt.title("Home Prices by Top 10 Zip Codes")
plt.tight_layout()
plt.ylabel('Price by Millions')
plt.xlabel('Zipcode')
```

These boxplots give us a good idea of the zipcodes themselves, but what about relative to each other? Here, we implement 3 chloroploths. Initially, we use seaborn for our implementation. We want to show price intensity with our measure of spread and central tendancy, median and mean respectively, and also measure the rate at which homes are sold in the top 10 zipcodes. We use a dictionary to map zipcode to these median, mean, and sales (houses sold), then mapping it to a dictionary.

```python
top10_zips_pre = pre_filtered['address/zipcode'].unique()
zip_price_info_pre = {}
for zipcode in top10_zips_pre:
    zip_price_info_pre[zipcode] = [pre_filtered[pre_filtered['address/zipcode'] == zipcode]['price'].median(), pre_filtered[pre_filtered['address/zipcode'] == zipcode]['price'].mean(), pre_filtered[pre_filtered['address/zipcode'] == zipcode]['price'].count()]
zip_price_info_pre_df = pd.DataFrame.from_dict(zip_price_info_pre, orient='index', columns=['median_price', 'mean_price', 'sales_count']).reset_index()
zip_price_info_pre_df = zip_price_info_pre_df.rename(columns={'index':'address/zipcode'})
zip_price_info_pre_df['address/zipcode'] = zip_price_info_pre_df['address/zipcode'].astype(str)
zip_price_info_pre_df
```

```python
top_zips_pass = pass_110['address/zipcode'].value_counts().head(10).index
def zipCompare(zipcodes_one, zipcodes_two):
    together = []
    separate_one = []
    separate_two = []
    for zipcode in zipcodes_one:
        if zipcode not in zipcodes_two:
            separate_one.append(zipcode)
        else:
            together.append(zipcode)
    for zipcode in zipcodes_two:
        if zipcode not in zipcodes_one:
            separate_two.append(zipcode)

    print('List of zipcodes in both top 10: ' + str(together))
    print('List of zipcodes in top 10 of list one: ' + str(separate_one))
    print('List of zipcodes in top 10 of list two: ' + str(separate_two))
    return together
        
    
shared_zips_pre_pass = zipCompare(top_zips_pre, top_zips_pass)
```

```python
top_zips_post = post_110['address/zipcode'].value_counts().head(10).index
def zipCompare2(zipcodes_one, zipcodes_two, zipcodes_three):
    together = []
    one = []
    two = []
    three = []
    one_and_two = []
    one_and_three = []
    two_and_three = []
    for zipcode in zipcodes_one:
        if zipcode in zipcodes_two and zipcode not in zipcodes_three:
            one_and_two.append(zipcode)
        elif zipcode not in zipcodes_two and zipcode in zipcodes_three:
            one_and_three.append(zipcode)
        elif zipcode not in zipcodes_two and zipcode not in zipcodes_three:
            one.append(zipcode)
        else:
            together.append(zipcode)
    for zipcode in zipcodes_two:
        if zipcode not in zipcodes_one and zipcode in zipcodes_three:
            two_and_three.append(zipcode)
        elif zipcode not in zipcodes_one and zipcode not in zipcodes_three:
            two.append(zipcode)
    for zipcode in zipcodes_three:
        if zipcode not in zipcodes_one and zipcode not in zipcodes_two:
            three.append(zipcode)

    print('List of zipcodes in both top 10: ' + str(together))
    print('List of zipcodes in just one: ' + str(one))
    print('List of zipcodes in just two: ' + str(two))
    print('List of zipcodes in just three: ' + str(three))
    print('List of zipcodes in one and two: ' + str(one_and_two))
    print('List of zipcodes in one and three: ' + str(one_and_three))
    print('List of zipcodes in two and three: ' + str(two_and_three))
    return together
        
    
shared_zips_pre_pass_post = zipCompare2(top_zips_pre, top_zips_pass, top_zips_post)
```

#### Analysis 4: Housing vs. Opioid Outcomes
We want to determine if the opioid crisis has a measurable economic footprint on local neighborhoods. We are testing whether high opioid incidents has negative impact on property values and market liquidity, specifically comparing the pre-to-post pandemic shifts between 2020 and 2021. AI was used to understand how to properly use spearman rho and to help figure out how to format the visuals to be less clunky.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats
# I want the figure to look nice and clean, so this is just some work to make sure it looks good.
sns.set_theme(style="whitegrid", context="talk")
plt.rcParams['figure.figsize'] = (16, 10)
```

Aggregating the data, I think it would be good to collapse individual house sales into a single county level representative metric. Since it's not possible to correlate a lot of housing prices to one death rate without many-to-one bias, we use median house price to create 1:1 relationship for the tests. 

```python
county_level_df = eda_df.groupby(['County', 'yearSold']).agg({
    'price': 'median',
    'daysOnZillow': 'median',
    'Rate (per 100K)': 'first'  # rate is already unique per county-year
}).reset_index()
```

We now create comparison groups. I will convert the continuous rates into 3 categories (Low/Medium/High), which lets us answer how the worst hit areas compare to least hit areas. 

```python
county_level_df['Opioid_Risk'] = pd.qcut(county_level_df['Rate (per 100K)'], 3, 
                                         labels=['Low Risk', 'Medium Risk', 'High Risk'])
```

I need to see if there is a consistent trend. I'm using sns.regplot to see the scatter points along with a regression line. The shaded area shows the 95% confidence interval, which helps me see how much "noise" or uncertainty is in the relationship.

```python
# create a 2x2 grid to compare Price and Time on the market across 2020 and 2021
fig, axes = plt.subplots(2, 2, figsize=(18, 12))
years = [2020, 2021]
metrics = [('price', 'Median Price ($)'), ('daysOnZillow', 'Days on Zillow')]

for r, year in enumerate(years):
    subset = county_level_df[county_level_df['yearSold'] == year]
    for c, (col_name, label) in enumerate(metrics):
        ax = axes[r, c]
        sns.regplot(data=subset, x=col_name, y='Rate (per 100K)', ax=ax, 
                    scatter_kws={'alpha':0.6}, line_kws={'color':'steelblue'})
        
        # Spearman Rho measures the rank order, telling us if higher rates 
        # consistently lead to lower prices, even if it's not a perfect straight line.
        corr, _ = stats.spearmanr(subset[col_name], subset['Rate (per 100K)'])
        ax.set_title(f"{year}: Opioid Rate vs {label}")
        ax.annotate(f'Spearman Rho: {corr:.2f}', xy=(0.05, 0.9), xycoords='axes fraction', 
                    fontsize=14, bbox=dict(boxstyle="round", fc="w"))

plt.tight_layout()
plt.show()

# create a clear summary table to validate the final hypothesis.
summary = county_level_df.groupby(['yearSold', 'Opioid_Risk'], observed=True)[['price', 'daysOnZillow']].median()
print(summary)
```

The regression plots for both years show a perfect negative Spearman Rho of -1.00. This indicates a consistent relationship that: as opioid incidence increases, housing valuations decrease without exception. In 2020, the difference in median price between low risk and high risk areas was 111,975 dollars. By 2021, even though prices rose across the board, the gap widened to $119,000. This suggests that while the 2021 housing boom lifted all markets, high-opioid areas did not benefit as significantly as lower risk neighborhoods.
The steep downward slope of the regression line suggests that opioid rates act as a price ceiling. Buyers appear to demand a significant price discount to move into areas with higher public health crises, regardless of the current economic climate.


The plots for market speed show a perfect positive Spearman Rho of 1.00. This confirms that higher opioid rates are a direct predictor of market stagnation. There appears to be a dramatic shift in the x axis from 2020 to 2021. In 2020, homes sat for nearly 294–295 days regardless of risk level. By 2021, the market accelerated greatly, with homes selling in roughly 77–83 days. Despite the massive market acceleration in 2021, the high risk areas still fell behind low risk areas by 6 days. While 6 days seems small in absolute terms, it represents nearly an 8% increase in time on market compared to low risk areas. 

To summarize, higher opioid-rate areas definitely tend to have lower prices and longer days on the market. This is made apparant in the stagnation and lower valuation of the local housing market.

This analysis shows the relationship between opioid overdose rates and local housing valuations. By regressing median house prices on mortality rates, the goal is to determine if higher incidence rates act as a "price ceiling" on the market, potentially causing stagnation in property values even during significant housing booms. Using heteroscedasticity-robust standard errors (HC3), we ensure that our findings are statistically grounded, even when dealing with the inherent volatility of real estate data across different counties and years. We moved from a simple bivariate model to a year-adjusted Model to isolate the "opioid effect" from the broader 2021 housing boom.

```python
import statsmodels.formula.api as smf

# 1. Clean the column names for statsmodels formula compatibility
# Statsmodels doesn't like spaces or special characters in names
reg_df = county_level_df.rename(columns={
    'Rate (per 100K)': 'opioid_rate',
    'price': 'median_price'
})

# 2. simple Regression with Heteroscedasticity-Robust Standard Errors (HC3)
# We use log(price) often in real estate to interpret the result as % change
simple_model = smf.ols('median_price ~ opioid_rate', data=reg_df).fit(cov_type='HC3')

# 3. Regression by Year (using fixed effects for yearSold)
# This checks if the 'opioid effect' changes when accounting for the 2021 boom
year_model = smf.ols('median_price ~ opioid_rate + C(yearSold)', data=reg_df).fit(cov_type='HC3')

print("--- Simple Model Summary ---")
print(simple_model.summary())

print("\n--- Year-Adjusted Model Summary ---")
print(year_model.summary())

```

In the adjusted model, the opioid_rate coefficient of -6,989.70 indicates that for every one-unit increase in the overdose rate per 100k residents, the median home price is expected to decrease by approximately 6,990 dollars. This suggests a kind of "financial discount" that the market applies to areas struggling with high overdose rates. Furthermore, while the model identified a significant upward shift in 2021, which adds  a baseline of roughly $79,710 to valuations regardless of location, the high R-squared of 0.958 confirms that the opioid rate remains a dominant predictor of how much of that "boom" a specific area actually captures. The shift in significance (improving from p=0.407 to p=0.073 after controlling for year) suggests that the opioid crisis acts as a constant downward pressure. Even when economic conditions are favorable, high incidence areas face a "ceiling" that prevents them from achieving the same growth as lower risk regions.
