## Background and Prior Work

Our topic is on housing prices & drug usage in Portland, specifically before and after the 2020 Oregon Ballot Measure 110 was passed. Measure 110 reduced the penalties for drug possession for small amounts of controlled substances, reducing it from a class A misdemeanor to class E. We want to study how this may have affected the general population of Portland's access to housing, and if drug usage was related during this time period. By 2024 measure 110 was largely repealed.<a name = "Drug_Decriminalization">[</a><sup>4</sup>](#Drug_Decriminalization_ref)

A number of studies were conducted about this time period, all which studied the effect of drug decriminalization in Portland in specific categories. 

One study found that housing instability and opioid related harms often rise together, especially in areas facing rapid policy shifts. Using cross-sectional survey data from unstably housed individuals with a history of drug use, the study identified key demographic and behavioral factors associated with access to housing assistance, including frequency of opioid use and community supervision status. The study stated that, "Participants were recruited [...] with partner agencies that provide supportive services [...] direct outreach by the study team to homeless"<a name="cite_ref-1"></a>[<sup>2</sup>](#cite_note-1). While this research focuses on individuals already experiencing housing instability, our study differs by examining broader population-level trends in Portland before and after the passage of Measure 110. Rather than relying on survey data, we analyze changes in housing prices alongside drug usage indicators to explore whether decriminalization coincided with shifts in housing accessibility and market conditions. This allows us to investigate potential associations at a city-wide level and assess how policy changes may have impacted both housing dynamics and drug use simultaneously over time.

Another study examined whether Oregon’s 2021 decriminalization of drug possession was linked to changes in overdose mortality while accounting for the spread of fentanyl in the unregulated drug market. The study found that “decriminalization of drug possession was not associated with an increase in fatal drug overdose rates in Oregon in the 2 years after its enactment” once fentanyl was considered as a confounding factor<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2). The authors emphasized that fentanyl is “the principal driver of the overdose mortality epidemic in the US,” suggesting that increases in overdose deaths cannot be attributed to Measure 110 alone. While this research focuses on overdose mortality rather than housing outcomes, it is still relevant to our study because it highlights the importance of accounting for external factors when evaluating the effects of drug policy changes. Our research builds on this idea by examining housing prices and drug usage trends in Portland before and after Measure 110 to better understand how policy changes may have impacted the broader community.

## Hypothesis


We hypothesize that there will be a negative correlation between opioid-related incidents and both median housing prices and market velocity (average time spent on Zillow) in Portland between 2020 and 2021. Specifically, as opioid overdoses and hospitalizations increased in certain zip codes, we predict that home sale prices in those areas decreased relative to the city average, while the "days on Zillow" significantly increased.

**Dataset #1: Portland Housing Prices & Sales (Zillow)**

- **Dataset Name:** Portland Housing Prices, Sales (Jul 2020–Jul 2021)
- **Link:** https://www.kaggle.com/datasets/threnjen/portland-housing-prices-sales-jul-2020-jul-2021

- **Dataset Name:** Annual opioid-related deaths by county (Oregon Injury Data)

- **Shortcomings:** County-level only (no ZIP); housing is ZIP-level, so a crosswalk is needed; low counts (1–4) are suppressed; Oregon Health Authority notes not all deaths receive toxicology testing, so counts may underrepresent true overdoses

**Combining the datasets:** Housing data is at ZIP level; overdose data is at county level. We use a county–ZIP crosswalk (`county_zip_crosswalk.csv`) to map each housing record's ZIP to its county, then join to county-year overdose counts/rates. This allows us to compare housing outcomes (price, days on market) by area with opioid death rates in the same county.

We first note how our data is presented. These are index/row based similarity, so 19309 NE Glisan St, for instance, was on Zillow for 25 days.
```python
housing_drop_cols = housing[['abbreviatedAddress','address/city','address/zipcode','bathrooms','bedrooms','livingArea','price']]
price_history_housing = housing_drop_cols.join(priceHistory, how='left', lsuffix='_left', rsuffix='_right')
price_history_housing
```

First, we investigate pre Measure 110. We will use the dateSold column as a means to obtain our pre-110 dataset, aptly named "pre_110". We will want to get summary statistics on the priceing of houses, as we as group by zipcodes to explore trends in certain subsects of Portland. To do this, we first convert the **object** datatype in dateSold to a **datetime** type. Then, we will extract the month and year of each **datetime**, check if it is in the pre Measure 110 range we want to dive into, add that row to our pre_110 dataframe, and generate summary statistics.

```python
price_history_housing['dateSold'] = pd.to_datetime(price_history_housing['dateSold'])
pre_110 = price_history_housing[(price_history_housing['dateSold'].dt.year == 2020) & (price_history_housing['dateSold'].dt.month < 12)]
print(pre_110.shape)
pre_110.head()
```

We're given 4700 datapoints to work with. On further examination, we see some zipcodes have a miniscule amount of datapoints related to them. We will only view the top 10 of these datapoints. Also, based on the range of price we had in our summary statistics, we will cap things to be only within 98% range of our data; this exludes the bottom 1%, and top 1%. Now, we will want to visualize the data for each of these zipcodes. We have chosen to do a box plot, as it captures the spread of the dataset the most, which is what we are invested in for comparing our ranges. We will also include mean as a measure of central tendancy by setting the showmeans flag to be true. It is almost sure that the mean will trend above the median price, since we have more outliers in the millions of dollars range, rather than, say, 5 digit or even 4 digit housing prices.

```python
lower = pre_110['price'].quantile(0.1)
upper = pre_110['price'].quantile(0.99)
plt.figure(figsize=(14,6))

top_zips_pre = pre_110['address/zipcode'].value_counts().head(10).index
pre_filtered = pre_110[pre_110['address/zipcode'].isin(top_zips_pre)]
pre_filtered = pre_filtered[(pre_filtered['price'] >= lower) & (pre_filtered['price'] <= upper)]

print(pre_filtered.shape)

sns.boxplot(x='address/zipcode', y='price', hue='address/zipcode', data=pre_filtered, showmeans=True, meanprops={'marker': 'o', 'markerfacecolor': 'white', 'markeredgecolor': 'black', 'markersize': '8'})

plt.title("Home Prices by Top 10 Zip Codes")
plt.tight_layout()
plt.ylabel('Price by Millions')
plt.xlabel('Zipcode')
```

These boxplots give us a good idea of the zipcodes themselves, but what about relative to each other? Here, we implement 3 chloroploths. Initially, we use seaborn for our implementation. We want to show price intensity with our measure of spread and central tendancy, median and mean respectively, and also measure the rate at which homes are sold in the top 10 zipcodes. We use a dictionary to map zipcode to these median, mean, and sales (houses sold), then mapping it to a dictionary.

```python
top10_zips_pre = pre_filtered['address/zipcode'].unique()
zip_price_info_pre = {}
for zipcode in top10_zips_pre:
    zip_price_info_pre[zipcode] = [pre_filtered[pre_filtered['address/zipcode'] == zipcode]['price'].median(), pre_filtered[pre_filtered['address/zipcode'] == zipcode]['price'].mean(), pre_filtered[pre_filtered['address/zipcode'] == zipcode]['price'].count()]
zip_price_info_pre_df = pd.DataFrame.from_dict(zip_price_info_pre, orient='index', columns=['median_price', 'mean_price', 'sales_count']).reset_index()
zip_price_info_pre_df = zip_price_info_pre_df.rename(columns={'index':'address/zipcode'})
zip_price_info_pre_df['address/zipcode'] = zip_price_info_pre_df['address/zipcode'].astype(str)
zip_price_info_pre_df
```

```python
top_zips_pass = pass_110['address/zipcode'].value_counts().head(10).index
def zipCompare(zipcodes_one, zipcodes_two):
    together = []
    separate_one = []
    separate_two = []
    for zipcode in zipcodes_one:
        if zipcode not in zipcodes_two:
            separate_one.append(zipcode)
        else:
            together.append(zipcode)
    for zipcode in zipcodes_two:
        if zipcode not in zipcodes_one:
            separate_two.append(zipcode)

    print('List of zipcodes in both top 10: ' + str(together))
    print('List of zipcodes in top 10 of list one: ' + str(separate_one))
    print('List of zipcodes in top 10 of list two: ' + str(separate_two))
    return together
        
    
shared_zips_pre_pass = zipCompare(top_zips_pre, top_zips_pass)
```

```python
top_zips_post = post_110['address/zipcode'].value_counts().head(10).index
def zipCompare2(zipcodes_one, zipcodes_two, zipcodes_three):
    together = []
    one = []
    two = []
    three = []
    one_and_two = []
    one_and_three = []
    two_and_three = []
    for zipcode in zipcodes_one:
        if zipcode in zipcodes_two and zipcode not in zipcodes_three:
            one_and_two.append(zipcode)
        elif zipcode not in zipcodes_two and zipcode in zipcodes_three:
            one_and_three.append(zipcode)
        elif zipcode not in zipcodes_two and zipcode not in zipcodes_three:
            one.append(zipcode)
        else:
            together.append(zipcode)
    for zipcode in zipcodes_two:
        if zipcode not in zipcodes_one and zipcode in zipcodes_three:
            two_and_three.append(zipcode)
        elif zipcode not in zipcodes_one and zipcode not in zipcodes_three:
            two.append(zipcode)
    for zipcode in zipcodes_three:
        if zipcode not in zipcodes_one and zipcode not in zipcodes_two:
            three.append(zipcode)

    print('List of zipcodes in both top 10: ' + str(together))
    print('List of zipcodes in just one: ' + str(one))
    print('List of zipcodes in just two: ' + str(two))
    print('List of zipcodes in just three: ' + str(three))
    print('List of zipcodes in one and two: ' + str(one_and_two))
    print('List of zipcodes in one and three: ' + str(one_and_three))
    print('List of zipcodes in two and three: ' + str(two_and_three))
    return together
        
    
shared_zips_pre_pass_post = zipCompare2(top_zips_pre, top_zips_pass, top_zips_post)
```

#### Analysis 4: Housing vs. Opioid Outcomes
We want to determine if the opioid crisis has a measurable economic footprint on local neighborhoods. We are testing whether high opioid incidents has negative impact on property values and market liquidity, specifically comparing the pre-to-post pandemic shifts between 2020 and 2021. AI was used to understand how to properly use spearman rho and to help figure out how to format the visuals to be less clunky.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats
# I want the figure to look nice and clean, so this is just some work to make sure it looks good.
sns.set_theme(style="whitegrid", context="talk")
plt.rcParams['figure.figsize'] = (16, 10)
```

```python
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf
import matplotlib.pyplot as plt
import seaborn as sns

#Create a dummy for year fixed effects
eda_df['is_2021'] = (eda_df['yearSold'] == 2021).astype(int)

#following the literature we work in log price
eda_df['log_price'] = np.log(eda_df['price'])

# log_price = B1 * OpioidRate + B2 * is_2021 + B3 * OpioidRate × is_2021 + u
model = smf.ols(
    formula='log_price ~ Q("Rate (per 100K)") * is_2021',
    data=eda_df
).fit(cov_type='HC1') # Use heteroskedasticity robust standard error WLOG

print(model.summary())
```

I think the first thing that jumps out to me from this regression is the very low R^2 of 0.018, indicating that the regression doesnt even explain 2 precent of the variation in price. This makes sense in general as we tend to think that most of the variation in house prices is related to size of the house, overall state of the housing market, etc. Of course, given our research question, it still is a little bit disappointing.

That being said, with the variation that we can account for the regression results indicate a clear negative relationship between opioid incidence and housing prices in 2020, where higher opioid rates are associated with lower home prices. Specifically, a one-unit increase in the opioid rate per 100K corresponds to approximately a 1.76% decrease in housing prices. In 2021, however, this negative relationship becomes noticeably weaker. The positive and significant interaction term shows that the strength of the association between opioid incidence and housing prices declines, with the estimated effect shrinking to about a 0.88% decrease in prices per unit increase in opioid rate. Both of these coeficients are statisticly significant at all commonly used signicance levels (in economics).

Next lets do formal test on the difference between the two statistics

```python
test = model.t_test('Q("Rate (per 100K)"):is_2021 = 0')
print(test)
```

As we can see, the 95 precent confidence interval does not include zero, so we can conclude there is a statistical difference at the 95 precent confidence level.

Finally we show the regression graph for ease of understanding the data.

```python
#The following graph was made with chatgtp

sns.set(style="whitegrid")

# Create x range for regression predictions
x_vals = np.linspace(
    eda_df['Rate (per 100K)'].min(),
    eda_df['Rate (per 100K)'].max(),
    100
)

# Extract coefficients from the model
b0 = model.params['Intercept']
b1 = model.params['Q("Rate (per 100K)")']
b2 = model.params['is_2021']
b3 = model.params['Q("Rate (per 100K)"):is_2021']

# Predicted log_prices
y_2020 = b0 + b1 * x_vals
y_2021 = b0 + b2 + (b1 + b3) * x_vals

# Convert to actual prices
y_2020_price = np.exp(y_2020)
y_2021_price = np.exp(y_2021)

# Determine zoom limits based on predicted lines
y_min = min(y_2020_price.min(), y_2021_price.min()) * 0.95
y_max = max(y_2020_price.max(), y_2021_price.max()) * 1.05

plt.figure(figsize=(10,6))

# Scatter points (optional, low alpha)
sns.scatterplot(
    data=eda_df,
    x='Rate (per 100K)',
    y='price',
    hue='yearSold',
    palette=sns.color_palette(n_colors=2),
    alpha=0.2,
    legend=False
)

# Regression lines using default seaborn colors
palette = sns.color_palette(n_colors=2)
plt.plot(x_vals, y_2020_price, color=palette[0], label='2020 Regression')
plt.plot(x_vals, y_2021_price, color=palette[1], linestyle='--', label='2021 Regression')

plt.xlabel('Opioid Rate (per 100K)')
plt.ylabel('Housing Price ($)')
plt.title('Zoomed Regression Lines: Housing Prices vs Opioid Rate')
plt.legend()

# Zoom in on regression range
plt.xlim(x_vals.min(), x_vals.max())
plt.ylim(y_min, y_max)

plt.show()
```
