# Research Question

How does the relationship between opioids incedance (overdoses, hospitalizations, check into rehab) in Portland and housing prices as well as average time spent on zillow before being sold change in 2020 to 2021?


## Background and Prior Work

Our topic is on housing prices & drug usage in Portland, specifically before and after the 2020 Oregon Ballot Measure 110 was passed. Measure 110 reduced the penalties for drug possession for small amounts of controlled substances, reducing it from a class A misdemeanor to class E. We want to study how this may have affected the general population of Portland's access to housing, and if drug usage was related during this time period. By 2024 measure 110 was largely repealed.<a name = "Drug_Decriminalization">[</a><sup>4</sup>](#Drug_Decriminalization_ref)

A number of studies were conducted about this time period, all which studied the effect of drug decriminalization in Portland in specific categories. 

One study found that housing instability and opioid related harms often rise together, especially in areas facing rapid policy shifts. Using cross-sectional survey data from unstably housed individuals with a history of drug use, the study identified key demographic and behavioral factors associated with access to housing assistance, including frequency of opioid use and community supervision status. The study stated that, "Participants were recruited [...] with partner agencies that provide supportive services [...] direct outreach by the study team to homeless"<a name="cite_ref-1"></a>[<sup>2</sup>](#cite_note-1). While this research focuses on individuals already experiencing housing instability, our study differs by examining broader population-level trends in Portland before and after the passage of Measure 110. Rather than relying on survey data, we analyze changes in housing prices alongside drug usage indicators to explore whether decriminalization coincided with shifts in housing accessibility and market conditions. This allows us to investigate potential associations at a city-wide level and assess how policy changes may have impacted both housing dynamics and drug use simultaneously over time.

Another study examined whether Oregon’s 2021 decriminalization of drug possession was linked to changes in overdose mortality while accounting for the spread of fentanyl in the unregulated drug market. The study found that “decriminalization of drug possession was not associated with an increase in fatal drug overdose rates in Oregon in the 2 years after its enactment” once fentanyl was considered as a confounding factor<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2). The authors emphasized that fentanyl is “the principal driver of the overdose mortality epidemic in the US,” suggesting that increases in overdose deaths cannot be attributed to Measure 110 alone. While this research focuses on overdose mortality rather than housing outcomes, it is still relevant to our study because it highlights the importance of accounting for external factors when evaluating the effects of drug policy changes. Our research builds on this idea by examining housing prices and drug usage trends in Portland before and after Measure 110 to better understand how policy changes may have impacted the broader community.

# Hypothesis


We hypothesize that there will be a negative correlation between opioid-related incidents and both median housing prices and market velocity (average time spent on Zillow) in Portland between 2020 and 2021. Specifically, as opioid overdoses and hospitalizations increased in certain zip codes, we predict that home sale prices in those areas decreased relative to the city average, while the "days on Zillow" significantly increased.

#### Section 3: Opioid data — pre vs post Measure 110

Compare opioid overdose deaths (Count and Rate per 100K) by county in 2020 (pre–Measure 110) vs 2021 (post–Measure 110). Measure 110 took effect in February 2021.

We first aggregate opioid deaths to one row per county-year. Each county has a single death count and rate for each year in our merged dataset.

```python
# Pivot: opioid deaths by county, 2020 vs 2021
opioid_pre_post = eda_df.groupby(['County', 'yearSold']).agg({'Count': 'first', 'Rate (per 100K)': 'first'}).reset_index()

pivot_count = opioid_pre_post.pivot(index='County', columns='yearSold', values='Count')
pivot_rate = opioid_pre_post.pivot(index='County', columns='yearSold', values='Rate (per 100K)')
print("Count (deaths):")
print(pivot_count)
print("\nRate (per 100K):")
print(pivot_rate)
```

In total across our sample counties, deaths increased from 2020 to 2021. The pivot tables above break this down by county; the summary below shows totals and mean rates.

```python
# Total deaths across our sample counties, by year
totals = opioid_pre_post.groupby('yearSold').agg({'Count': 'sum', 'Rate (per 100K)': 'mean'}).round(1)
totals.columns = ['Total deaths', 'Mean rate (per 100K)']
print(totals)
```

**Count** is the raw number of opioid overdose deaths. **Rate (per 100K)** adjusts for population size, so we can compare counties fairly—a small county with 10 deaths may have a higher rate than a large county with 50 deaths.

```python
# Bar chart: opioid death rate by county, 2020 (pre) vs 2021 (post)
pivot_rate.plot(kind='bar', figsize=(10, 5))
plt.title("Opioid Death Rate (per 100K) by County: 2020 (Pre–Measure 110) vs 2021 (Post–Measure 110)")
plt.ylabel("Rate per 100K")
plt.xlabel("County")
plt.xticks(rotation=45, ha='right')
plt.legend(title='Year')
plt.tight_layout()
plt.show()
```

The chart above shows death rate by county. Higher rates indicate more overdose deaths per capita. Multnomah County (Portland) dominates our housing sample.

Raw death counts are shown below. Counties with larger populations naturally have higher counts, so Multnomah stands out. The rate (per 100K) is often more informative for comparing burden across counties of different sizes.

```python
# Bar chart: raw death count by county, 2020 vs 2021
pivot_count.plot(kind='bar', figsize=(10, 5))
plt.title("Opioid Death Count by County: 2020 (Pre–Measure 110) vs 2021 (Post–Measure 110)")
plt.ylabel("Number of deaths")
plt.xlabel("County")
plt.xticks(rotation=45, ha='right')
plt.legend(title='Year')
plt.tight_layout()
plt.show()
```

To quantify year-over-year change, we compute the absolute and percent change in death rate from 2020 to 2021 for each county. Positive values indicate an increase in the rate after Measure 110 took effect.

```python
# Year-over-year change in opioid death rate
change_df = pivot_rate.copy()
change_df['abs_change'] = change_df[2021] - change_df[2020]
change_df['pct_change'] = 100 * (change_df[2021] - change_df[2020]) / change_df[2020]
change_df = change_df[['abs_change', 'pct_change']].round(1)
change_df.columns = ['Rate change (2021−2020)', '% change']
print(change_df)
```

A bar chart of the percent change makes it easy to see which counties had the largest relative increase in opioid death rates from 2020 to 2021.

```python
# Bar chart: percent change in death rate, 2020 → 2021
change_df['% change'].plot(kind='bar', figsize=(10, 5), color='steelblue', edgecolor='navy')
plt.title("% Change in Opioid Death Rate (2020 → 2021) by County")
plt.ylabel("% change")
plt.xlabel("County")
plt.xticks(rotation=45, ha='right')
plt.axhline(y=0, color='gray', linestyle='--')
plt.tight_layout()
plt.show()
```

```python
county_level_df['Opioid_Risk'] = pd.qcut(county_level_df['Rate (per 100K)'], 3, 
                                         labels=['Low Risk', 'Medium Risk', 'High Risk'])
```

I need to see if there is a consistent trend. I'm using sns.regplot to see the scatter points along with a regression line. The shaded area shows the 95% confidence interval, which helps me see how much "noise" or uncertainty is in the relationship.

```python
# create a 2x2 grid to compare Price and Time on the market across 2020 and 2021
fig, axes = plt.subplots(2, 2, figsize=(18, 12))
years = [2020, 2021]
metrics = [('price', 'Median Price ($)'), ('daysOnZillow', 'Days on Zillow')]

for r, year in enumerate(years):
    subset = county_level_df[county_level_df['yearSold'] == year]
    for c, (col_name, label) in enumerate(metrics):
        ax = axes[r, c]
        sns.regplot(data=subset, x=col_name, y='Rate (per 100K)', ax=ax, 
                    scatter_kws={'alpha':0.6}, line_kws={'color':'darkred'})
        
        # Spearman Rho measures the rank order, telling us if higher rates 
        # consistently lead to lower prices, even if it's not a perfect straight line.
        corr, _ = stats.spearmanr(subset[col_name], subset['Rate (per 100K)'])
        ax.set_title(f"{year}: Opioid Rate vs {label}")
        ax.annotate(f'Spearman Rho: {corr:.2f}', xy=(0.05, 0.9), xycoords='axes fraction', 
                    fontsize=14, bbox=dict(boxstyle="round", fc="w"))

plt.tight_layout()
plt.show()

# create a clear summary table to validate the final hypothesis.
summary = county_level_df.groupby(['yearSold', 'Opioid_Risk'], observed=True)[['price', 'daysOnZillow']].median()
print(summary)
```

The regression plots for both years show a perfect negative Spearman Rho of -1.00. This indicates a consistent relationship that: as opioid incidence increases, housing valuations decrease without exception. In 2020, the difference in median price between low risk and high risk areas was $111,975. By 2021, even though prices rose across the board, the gap widened to $119,000. This suggests that while the 2021 housing boom lifted all markets, high-opioid areas did not benefit as significantly as lower risk neighborhoods.
The steep downward slope of the regression line suggests that opioid rates act as a price ceiling. Buyers appear to demand a significant price discount to move into areas with higher public health crises, regardless of the current economic climate.


The plots for market speed show a perfect positive Spearman Rho of 1.00. This confirms that higher opioid rates are a direct predictor of market stagnation. There appears to be a dramatic shift in the x axis from 2020 to 2021. In 2020, homes sat for nearly 294–295 days regardless of risk level. By 2021, the market accelerated greatly, with homes selling in roughly 77–83 days. Despite the massive market acceleration in 2021, the high risk areas still fell behind low risk areas by 6 days. While 6 days seems small in absolute terms, it represents nearly an 8% increase in time on market compared to low risk areas. 

To summarize, higher opioid-rate areas definitely tend to have lower prices and longer days on the market. This is made apparant in the stagnation and lower valuation of the local housing market.
